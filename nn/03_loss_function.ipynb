What a loss function is

So far our network produces an output, but it has no idea if that output is "right." A loss function compares the network's output to what it should have been (the target/label) and returns a single number representing how wrong it was. That number is what backpropagation will later try to shrink.

We'll use the simplest one: Mean Squared Error (MSE) — take the difference between predicted and actual, square it (so negatives don't cancel positives), and average.

```python
import torch 
import torch.nn as nn
```

```python
predicted = torch.tensor([0.5, 1.5, 1.0]) # what our layer produced last time
target = torch.tensor([1.0, 1.0, 1.0]) # what we WANTED it to produce 

# MSE by hand: for eahh pair, subtract, square, then average all of them
errors = predicted - target
squared_errors = errors ** 2
mse_by_hand = squared_errors.mean()

print("errors:", errors)
print("squared errors:", squared_errors)
print("MSE:", mse_by_hand)
```

```python
loss_fn = nn.MSELoss()
mse_bultin = loss_fn(predicted, target)

print("MSE (built-in):", mse_bultin)
```

A loss function measures how wrong the network's output is, compared to what it should have been (the target).

- Computed **Mean Squared Error** by hand: `((predicted - target) ** 2).mean()`
- Confirmed PyTorch's built-in `nn.MSELoss()` gives the exact same result
- This single number is what backpropagation will try to minimize by adjusting weights and biases
